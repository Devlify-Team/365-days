# Jupyter Notebook Best Practices

A reusable checklist for clear, reviewable notebooks. This starter contains no dataset and performs no analysis.

## 1. Start with a question

Write the purpose, audience, and one question before coding. A notebook is a narrative: pair each meaningful step with a short Markdown explanation of what it does and why. Keep it focused; move reusable logic into a tested Python module rather than duplicating large blocks across cells.

## 2. Record data provenance and permission

Before loading any real data, document its source, owner, access date, license or terms, and sensitive fields. Confirm that you are authorized to read, process, store, and share it. Keep credentials outside notebooks, and never commit private data or secrets. This exercise supplies no sample or synthetic dataset.

## 3. Make execution order obvious

A kernel keeps variables between cell runs. Re-running cells out of order can leave hidden state and misleading results. Keep cells small and top-to-bottom, name variables clearly, avoid unnecessary side effects, and restart the kernel then run all cells before sharing.

## 4. Keep inputs and assumptions explicit

Use project-relative paths; do not depend on an undocumented current directory or a private mount. Preserve raw data, document transformations, and validate assumptions before analysis. Uncomment and adapt the next cell only after documenting an authorized input file and its use restrictions.

## 5. Work in small, testable steps

For each code section, say what you expect to learn, inspect inputs and outputs for problems, and separate data cleaning, analysis, and presentation. Avoid huge tables, unbounded logs, and computations with unclear costs. When randomness is actually part of your method, record the seed and explain why.

## 6. Explain results and limitations

Summarize the result in plain language, state what evidence supports it, and note missing data, uncertain assumptions, and limitations. If a result matters, verify it with an independent check or move the calculation into tested code. Never present an unvalidated output as a conclusion.

## 7. Review before committing or sharing

Restart the kernel and run all cells from a fresh state. Remove debugging cells, clear unnecessary outputs, and inspect the saved notebook for secrets, private paths, personal information, large outputs, or licensed material you cannot share. Review the Git diff; notebook files contain code and may also contain saved outputs and comments. In Colab, sharing the notebook shares its contents but not the runtime machine, installed libraries, or files. Set sharing permissions intentionally.

## Ready-to-share checklist

- [ ] The question and intended audience are clear.
- [ ] Data source, access date, and permissions are documented.
- [ ] A fresh kernel can run the notebook top-to-bottom.
- [ ] Reusable code and assumptions are easy to find and test.
- [ ] Results have context, validation, and limitations.
- [ ] Outputs and sharing permissions have been reviewed.
